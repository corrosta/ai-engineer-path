# Мини-проект: Трекер обучения
Использует: списки, словари, циклы, функции, условия

```python
learning_log = []

record = {
    "day": 1,
    "topic": "Циклы и функции",
    "hours_spent": 3,
    "difficulty": "средне"
}

learning_log.append(record)

def add_entry(day, topic, hours_spent, difficulty):
    entry = {
        "day": day ,
        "topic": topic,
        "hours_spent": hours_spent,
        "difficulty": difficulty
    }
    learning_log.append(entry)
    print("Добавление новой записи в журнал обучения", day) 
```

```python
add_entry(2, "Циклы и функции", 3, "средне")
add_entry(3, "Мини-проекты", 0.5, "сложно")
```

```python
for entry in learning_log:
    print("День:", entry["day"])
    print("Тема:", entry["topic"])
    print("Время, потраченное на изучение:", entry["hours_spent"], "часов")
    print("Сложность:", entry["difficulty"])
    print()
```

```python
def show_stats():
    hours_spent = 0
    days_spent = len(learning_log)

    for entry in learning_log:
        hours_spent += entry["hours_spent"]
    return hours_spent, days_spent

print("Общее количество часов, потраченных на обучение:", show_stats()[0])
print("Общее количество дней, потраченных на обучение:", show_stats()[1])
```

```python
difficulty_levels = {
    "легко": 1,
    "средне": 2,
    "сложно": 3
}

entry = learning_log[0]
difficulty_text = entry["difficulty"].lower()
print(difficulty_text)
```

```python
def hardest_day():
    hardest_entry = None
    hardest_score = 0

    for entry in learning_log:
        current_difficulty = entry["difficulty"].lower()
        current_score = difficulty_levels[current_difficulty]
        if current_score > hardest_score: 
            hardest_score = current_score
            hardest_entry = entry

    return hardest_entry

print(hardest_day())
```
