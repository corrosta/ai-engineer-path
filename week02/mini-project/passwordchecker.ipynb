```python
#Введите Ваше имя:
name = "corrosta"
#Введите Ваш пароль:
password = "sSs222!" #Ваш пароль должен содержать минимум 8, 1 заглавную букву и 1 символ

def check_password(password):
    alpha = False
    digital_coant = 0
    upper_coant = 0
    sym_coant = 0

    for char in password :
        if char.isdigit():
            digital_coant +=1
        if char.isalpha() :
            alpha = True
        if char.isupper() :
            upper_coant +=1
        if not char.isalpha() and not char.isdigit() :
            sym_coant +=1


    problems = []

    if len(password) < 8 :
        problems.append("слишком короткий")
    if digital_coant < 2 :
        problems.append("имеет меньше 2-ух цифр")
    if sym_coant < 1 :
        problems.append("не имеет хотя бы одного символа")
    if not alpha:
        problems.append("не имеет букв")
    if upper_coant < 1:
        problems.append("не имеет заглавную букву")

    if len(problems) == 0:
       return("Надёжный, спасибо!")
    else:
        return("Слабый: " + ", ".join(problems) + ". Повторите попытку!")

result = check_password(password)

print(f"Здравствуйте, {name}, Ваш пароль: {result}")
```
