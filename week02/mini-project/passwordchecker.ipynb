```python
#Введите Ваш пароль:
password = "222222"

def check_password(password):
    digitals = False
    alpha = False

    for char in password :
        if char.isdigit():
            digitals = True
        if char.isalpha() :
            alpha = True

    problems = []

    if len(password) < 8 :
        problems.append("слишком короткий")
    if not digitals:
        problems.append("не имеет цифр")
    if not alpha:
        problems.append("не имеет букв")

    if len(problems) == 0:
       return("Надёжный, спасибо!")
    else:
        return("Слабый: " + ", ".join(problems))

print("Ваш пароль:" , check_password(password))
```
