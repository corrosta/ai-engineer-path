```python
#Введите Ваше имя:
login = "corrosta"

#Введите Ваш e-mail:
email = "test@test.ru"

#Введите Ваш пароль:
password = "777SSSSSS!" #Ваш пароль должен содержать минимум 8, 1 заглавную букву и 1 символ


#Проверка пароля
def check_password(password):
        reg_problems = []
        problems = []
        alpha = False
        digital_coant = 0
        upper_coant = 0
        sym_coant = 0

        for char in password :
            if char.isdigit():
                digital_coant +=1
            if char.isalpha() :
                alpha = True
            if char.isupper() :
                upper_coant +=1
            if not char.isalpha() and not char.isdigit() :
                sym_coant +=1

        if len(password) < 8 :
            problems.append("слишком короткий")
        if digital_coant < 2 :
            problems.append("имеет меньше 2-ух цифр")
        if sym_coant < 1 :
            problems.append("не имеет хотя бы одного символа")
        if not alpha:
            problems.append("не имеет букв")
        if upper_coant < 1:
            problems.append("не имеет заглавную букву")
        if len(problems) >= 1:
            return(", ".join(problems) + ". Повторите попытку!")
        else: return("Надёжный, спасибо!")

users_db = []

def register_user(login,email,password):
    result = check_password(password)
    reg_problems = []
    for user_entry in users_db: 
            if user_entry["login"] == login and user_entry["password"] == password:
                    reg_problems.append("Пользователь уже зарегистрирован,попробуйте войти")
                    break
            if user_entry["email"] == email:
                            reg_problems.append("Пользователь с таким email уже зарегистрирован")
                            break
    if result == "Надёжный, спасибо!" and len(reg_problems) == 0:
            new_user = {
                "login": login,
                "email": email,
                "password": password
            }
            users_db.append(new_user)
            print(f"Пользователь {login} успешно зарегистрирован!")
    else: 
            print(f"При регистрации {login} возникли ошибки:")
            for problems in reg_problems:
                print(problems)
            if result != "Надёжный, спасибо!":
                 print("Пароль слишком слабый:" + result)
                 

register_user(login,email,password)

#Тест дубликата почты
register_user("Azat", "test@test.ru", "SSSKKKasdhasd612@@")

#Тест пароля
register_user("KKa", "test3@test.ru", "k22222222222")

#Тест Нового пользователя
register_user("Marat","marat@test.ru","MaratStrong1337!")

#Тест Зарегистрированного пользователя
register_user(login,email,password)

```
